## <font color='#34cd83' face='Courier New'>DBSCAN</font>

    Inisialisasi parameter
        ↓       
    Penentuan tetangga (pengukuran jarak data dengan data yang lain)
        ↓       
    Pembentukan cluster setelah jarak diukur
        ↓       
    Iterasi dilakukan terhadap semua titik di dalam dataset

## <font color='#34cd83' face='Courier New'>PARAMETER</font>

### <font color='#34cd83' face='Courier New'>epsilon</font>
<div style="color:#ffffff; font-family:Courier New;">
    Jarak maksimum yang digunakan untuk menentukan bahwa dua titik dianggap saling berdekatan <br>
</div>

### <font color='#34cd83' face='Courier New'>min_sample</font>
<div style="color:#ffffff; font-family:Courier New;">
    jumlah titik minimum yang diperlukan di dalam lingkaran untuk suatu kumpulan data dianggap cluster <br>
</div>
