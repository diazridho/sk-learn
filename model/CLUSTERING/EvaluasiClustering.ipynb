## <font color='#34cd83' face='Courier New'>Silhouette Score</font>
<div style="color:#ffffff; font-family:Courier New;">
    Metrics yang mengukur seberapa baik suatu data point diklasifikasikan dalam <br>
    clusternya sendiri dibanding dengan cluster lain <br> <br>
    1: data point diklasifikasikan dengan baik di cluster sendiri dan terpisah dengan baik dari cluster lain <br>
    -1: data point lebih cocok berada di cluster lain
</div>

## <font color='#34cd83' face='Courier New'>Within cluster sum of square (WCSS)</font>
<div style="color:#ffffff; font-family:Courier New;">
    Penjumlahan jarak kuadrat antara setiap data point dengan centroidnya <br>
</div>
