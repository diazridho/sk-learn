## <font color='#34cd83' face='Courier New'>KMeans</font>

<div style="color:#ffffff; font-family:Courier New;">
    Metode clustering yang membagi data ke dalam k cluster berdasarkan kedekatan data dengan centroid <br>
</div>

    Menentukan jumlah cluster
        ↓       
    Inisialisasi centroid (dilakukan secara acak)
        ↓       
    Pengalokasian setiap data terhadap centroidnya dan menjadi sebuah cluster
        ↓       
    Menghitung ulang centroid dari cluster yang didapat sebelumnya
        ↓       
    Pengalokasian kembali data terhadap centroid baru 
        ↓       
    Iterasi sampai didapatkan centroid yang stabil

## <font color='#34cd83' face='Courier New'>Ciri-ciri</font>
<div style="color:#ffffff; font-family:Courier New;">
    1. Sensitif terhadap outlier <br>
    2. Untuk data numerik <br>
    3. Scalable
</div>
