## <font color='#cd9a34' face='Courier New'>Parameter KNN</font>

### <font color='#cd9a34' face='Courier New'>Jumlah Tetangga(K, default=5)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Menentukan jumlah tetangga yang dipertimbangkan untuk prediksi <br> <br>
</div>
<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    &nbsp;&nbsp;&nbsp;&nbsp;- Semakin sedikit, maka pengaruh tetangga semakin besar. Menyebabkan sensitif terhadap outliers dan overfitting <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Semakin banyak, maka pengaruh antar tetangga akan semakin terbagi. Menyebabkan underfitting
</div>

### <font color='#cd9a34' face='Courier New'>Metric Jarak (default = minkowski, p = 2)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Mengukur seberapa dekat suatu data dengan data yang berlabel. Hal ini memengaruhi pemilihan tetangga terdekat.<br>
    <br>
</div>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    Jenis metrix KNN: <br>
    &nbsp;&nbsp;&nbsp;&nbsp;-   Euclidean Distance(mengukur garis lurus) akar (p-q)**2 <br>
    &nbsp;&nbsp;&nbsp;&nbsp;-   Manhattan Distance(mengukur jumlah perbedaan di koordinat) absolut(p-q) <br>
    &nbsp;&nbsp;&nbsp;&nbsp;-   Minkowski Distance(parameter p= 1 akan menjadi Manhatan, p=2 akan menjadi Euclidean) <br>
    &nbsp;&nbsp;&nbsp;&nbsp;-   Cosine Similarity(kesamaan sudut antar dua vektor, untuk klasifikasi teks)
</div>

### <font color='#cd9a34' face='Courier New'>Weights</font>

<div style="color:#ffffff; font-family:Courier New;">
    Weights (Bobot Pengaruh):
    <ul style="list-style-type: circle; color: #fffff;">
        <li>Uniform: Pengaruh tetangga sama semua</li>
        <li>Distance: Semakin dekat, semakin besar pengaruhnya</li>
    </ul>
</div>

### <font color='#cd9a34' face='Courier New'>Algoritma pencari tetangga (default=auto)</font>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    1. Brute Force: Menghitung jarak antara data baru dan setiap titik dalam dataset pelatihan secara langsung. Lambat untuk dataset besar karena kompleksitas waktu yang tinggi. <br>
    2. KD-Tree: Pemotongan terhadap setiap fitur dan dilakukan berulang. Untuk dataset berdimensi kecil atau sedang <br>
    3. Ball-Tree: Mencari tetangga dengan membungkus data dalam bola. Untuk data berdimensi tinggi.  
</div>


## <font color='#cd9a34' face='Courier New'>Ciri khas KNN</font>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    1. Tidak mengasumsikan distribusi <br>
    2. Lambat dan memakan memori untuk dataset besar, idealnya kurang dari 20 fitur dan 50.000-100.000 baris <br>
    3. Penting untuk di skalarisasi
</div>
