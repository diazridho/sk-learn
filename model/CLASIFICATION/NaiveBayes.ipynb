## <font color='#c8cd34' face='Courier New'>Parameter Naive Bayes</font>

### <font color='#c8cd34' face='Courier New'>alpha(default=1.0) atau laplace smoothing</font>

<div style="color:#ffffff; font-family:Courier New;">
    Metode pencegahan untuk P(E|H) jika E nol, menambahkan 1 ke pembilang dan menambahkan V ke penyebut <br>
</div>

### <font color='#c8cd34' face='Courier New'>Probabilitas Prior</font>
<div style="color:#ffffff; font-family:Courier New;">
    Probabilitas munculnya suatu kelas pada target <br>
</div>

### <font color='#c8cd34' face='Courier New'>Probabilitas Likelihood/Bersyarat</font>
<div style="color:#ffffff; font-family:Courier New;">
    Probabilitas munculnya suatu kejadian jika kelas tertentu terjadi P(E|H) <br>
</div>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    1. KNN (K Nearest Neighbors) <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Angka Desimal (Kontinu) GaussianNB <br>
    2. Decision Tree <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Frekuensi Kata / Hitungan MultinomialNB <br>
    3. Random Forest <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Label (Ada/Tidak Ada) BernoulliNB  <br>
    4. Logistic Regression <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Kategori (Cerah, Hujan, dll) CategoricalNB<br>
</div>
