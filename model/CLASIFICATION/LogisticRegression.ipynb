## <font color='#cd348b' face='Courier New'>Logistic Regression</font>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">   
1. Binary LR: Dependent variabel punya dua kategori (yes/no, 0/1) <br>
2. Multinomial LR: y punya lebih dari dua kategori tanpa urutan (red/blue/green) <br>
3. Ordinal LR: y punya lebih dari dua kategori dengan urutan (low/medium/high)
</div>

### <font color='#cd348b' face='Courier New'>Ciri-ciri Logistic Regression</font>

<div style="color:#ffffff; font-family:Courier New;">
    1. Mengasumsikan tidak ada outlier <br>
    2. Sample besar <br>
</div>

### <font color='#cd348b' face='Courier New'>Binomial LogisticRegression</font>

<div style="color:#ffffff; font-family:Courier New;">
    Sigmoid function: <br>
    Mengubah persamaan linear menjadi probabilitas antara 0-1 <br> <br>

    Input X     
        ↓       
    Persamaan linear (wx + b) 
        ↓       
    Sigmoid function 
        ↓       
    Probabilitas kelas positif
        ↓       
    Threshold   
        ↓       
    Prediksi kelas
</div>

### <font color='#cd348b' face='Courier New'>Multinomial LogisticRegression</font>

<div style="color:#ffffff; font-family:Courier New;">
    Softmax function: <br>
    Mengubah hasil persamaan linear menjadi probabilitas yang jumlahnya 1<br> <br>

    Input X 
        ↓   
    Persamaan linear untuk setiap kelas terhadap fitur 
        ↓   
    Softmax function (e^z masing-masing kelas / Σe^z semua kelas)
        ↓   
    Probabilitas untuk setiap kelas
        ↓   
    Prediksi kelas dengan probabilitas tertinggi
</div>

### <font color='#cd348b' face='Courier New'>Ordinal LogisticRegression</font>

<div style="color:#ffffff; font-family:Courier New;">

    Input X 
        ↓   
    Hitung skor laten z = wᵀx
        ↓   
    Bandingkan z dengan setiap threshold
        ↓   
    Sigmoid → probabilitas melewati tiap level
        ↓   
    Hitung probabilitas tiap kelas (selisih kumulatif)
        ↓   
    Pilih kelas dengan probabilitas terbesar
</div>
