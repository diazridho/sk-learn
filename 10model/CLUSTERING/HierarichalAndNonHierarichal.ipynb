## <font color='#c8cd34' face='Courier New'>Hierarichal Clustering</font>

<div style="color:#ffffff; font-family:Courier New;">
    Teknik pengelompokan data dalam bentuk hierarki <br>
</div>

### <font color='#c8cd34' face='Courier New'>Agglomerative (Bottom-up)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Pendekatan clustering dimulai dari setiap objek yang dianggap sebagai cluster individu <br>
</div>

### <font color='#c8cd34' face='Courier New'>Divisive (Top-down)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Pendekatan clustering dimulai dari satu cluster besar yang mencakup semua objek <br>
</div>
