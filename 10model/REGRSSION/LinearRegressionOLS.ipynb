## <font color='#c8cd34' face='Courier New'>Linear Regression OLS</font>

<div style="color:#ffffff; font-family:Courier New;">
    Algoritma yang berusaha mencari best fit line untuk memodelkan hubungan antara variabel independent dengan dependent <br>
</div>

### <font color='#c8cd34' face='Courier New'>Parameter utama</font>

<div style="color:#ffffff; font-family:Courier New;">
    - Intercept, nilai y ketika seluruh input x adalah 0 <br>
    - Koefisien/gradient, seberapa nilai y berubah ketika nilai x berubah <br>
    - Residual, y asli dikurangi y prediksi

</div>

### <font color='#c8cd34' face='Courier New'>Algoritma OLS (Mencari normal equation)</font>

<div>

    Mendefinisikan matriks X, dengan memberikan angka 1 di kolom pertama untuk intercept
        ↓       
    Mencari matriks(Xtranspose . X)
        ↓       
    Invers matriks di atas
        ↓       
    Mencari matriks (Xtranspose . y)
        ↓       
    Mencari normal equation [invers(Xtranspose . X) . (Xtranspose . y)]
        ↓       
    Normal equation dijadikan persamaan linear/polynomial

</div>
