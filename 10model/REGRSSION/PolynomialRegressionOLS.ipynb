### <font color='#c8cd34' face='Courier New'>Algoritma OLS (Mencari normal equation)</font>


    Feature Engineering, menambahkan fitur sesuai degree yang dipilih.
    Contoh degree 2. X₁  X₂ akan ditambahkan X₁²   X₂²  X₁X₂
        ↓  
    Mendefinisikan matriks X, dengan memberikan angka 1 di kolom pertama untuk intercept
        ↓       
    Mencari matriks(Xtranspose . X)
        ↓       
    Invers matriks di atas
        ↓       
    Mencari matriks (Xtranspose . y)
        ↓       
    Mencari normal equation [invers(Xtranspose . X) . (Xtranspose . y)]
        ↓       
    Normal equation dijadikan persamaan linear/polynomial
