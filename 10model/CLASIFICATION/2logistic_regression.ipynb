```python
Binomial Logistic Regression

# Sigmoid function:
# Mengubah persamaan linear menjadi probabilitas antara 0-1

# Rumus sigmoid:
# σ(z) = 1 / (1 + e^(-z))
# dimana z = b0 + b1*x1 + b2*x2 + ... + bn*xn (persamaan linear)

# Input X
# ↓
# Persamaan linear (wx + b) 
# ↓
# Sigmoid function 
# ↓
# Probabilitas kelas positif
# ↓
# Threshold
# ↓
# Prediksi kelas
```

```python
Multinomial Logistic Regression

# Softmax function

# Input X
# ↓
# Persamaan linear untuk setiap kelas terhadap fitur 
# ↓
# Softmax function (e^z masing-masing kelas / Σe^z semua kelas)
# ↓
# Probabilitas untuk setiap kelas
# ↓
# Prediksi kelas dengan probabilitas tertinggi
```

```python
Ordinal Logistic Regression

# Input X
#  ↓
# Hitung skor laten z = wᵀx
#  ↓
# Bandingkan z dengan setiap threshold
#  ↓
# Sigmoid → probabilitas melewati tiap level
#  ↓
# Hitung probabilitas tiap kelas (selisih kumulatif)
#  ↓
# Pilih kelas dengan probabilitas terbesar
```
