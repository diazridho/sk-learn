## <font color='#ff00ff'>Parameter Random Forest</font>

### <span style="color:#00eeff; font-family:Helvetica;">1. n_estimators</span> <span style="color:#888888; font-family:Helvetica; font-size:0.8em;">(default=100)</span>
<p style="color:#cccccc; font-family:Arial; font-size:14px; margin-left:20px;">
    Menentukan jumlah pohon (trees) yang akan dibangun dalam forest. Semakin tinggi jumlahnya, model akan semakin stabil namun proses training akan lebih lama.
</p>

Menentukan banyak decision tree yang akan dibuat. Pengambilan sampel data dilakukan dengan acak, mungkin untuk diambil data yang sama. Data yang tidak ikut dalam pengambilan sampel disebut Out Of Bag
