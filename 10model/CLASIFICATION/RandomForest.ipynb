## <font color='#cebcbc' face='Courier New'>Parameter Random Forest</font>

### <font color='#cebcbc' face='Courier New'> n_estimators(default=100)</font>
<div style="color:#ffffff; font-family:Courier New;">
    Menentukan jumlah pohon (trees) yang akan dibangun dalam forest. Semakin tinggi jumlahnya, model akan semakin stabil namun proses training akan lebih lama.
</div>

### <font color='#cebcbc' face='Courier New'> Proses</font>
<div style="color:#ffffff; font-family:Courier New;">
    
    Menentukan banyak decision tree yang akan dibuat. 
    Pengambilan sampel data dilakukan dengan acak, mungkin untuk diambil data yang sama.
    Data yang tidak ikut dalam pengambilan sampel disebut Out Of Bag.
</div>

### <font color='#cebcbc' face='Courier New'> Ciri-ciri</font>
<div style="color:#ffffff; font-family:Courier New;">
    Cocok untuk dataset yang memiliki banyak fitur, karena otomatis melakukan feature selection pada setiap sub pohon.
</div>
