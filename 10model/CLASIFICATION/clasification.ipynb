## <font color='#c8cd34' face='Courier New'>ALGORITMA KLASIFIKASI</font>

1. KNN (K Nearest Neighbors)
    - Mengklasifikasikan data berdasarkan kedekatan dengan data yang sudah diberi label. Terlihat pengelompokan data
2. Decision Tree
    - Dapat mengelola data tidak terstruktur, baseline dari pohon struktur, lebih mudah divisualisasikan, tidak mengasumsikan distribusi/skala
3. Random Forest
    - Versi lanjutan dari decision tree
4. Logistic Regression
    - Mengubah persamaan linear menjadi probabilitas, untuk klasifikasi binner (Ya/Tidak)
5. Support Vector Machine 
    - Mencoba menemukan garis pemisah terbaik untuk membagi data ke kelas tertentu, efektif untuk data yang terpisah jelas, untuk klasifikasi binner/multikelas
6. Naive Bayes
    - Algoritma yang didasari Teorema Bayes, kemungkinan setiap fitur muncul dalam suatu kelas, untuk klasifikasi teks, banyak fitur kategori
