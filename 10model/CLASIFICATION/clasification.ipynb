## <font color='#00FFFF'>CLASIFICATION</font>

1. Klasifikasi Binner
2. Klasifikasi Multiclass
3. Klasifikasi Multilabel (Data memiliki lebih dari satu label)

### <font color='#00FFFF'>Algoritma Klasifikasi</font>

1. KNN (K Nearest Neighbors)
    - Mengklasifikasikan data berdasarkan kedekatan dengan data yang sudah diberi label.
2. Decision Tree
    - Dapat mengelola data tidak terstruktur, baseline dari pohon struktur, lebih mudah divisualisasikan, tidak perlu normalisasi/standardisasi
3. Random Forest
    - Versi lanjutan dari decision tree
4. Logistic Regression
    - Mengubah persamaan linear menjadi probabilitas, untuk klasifikasi binner
5. Support Vector Machine 
    - Mencoba menemukan garis pemisah terbaik untuk membagi data ke kelas tertentu, efektif untuk data yang terpisah jelas, untuk klasifikasi binner/multikelas
6. Naive Bayes
    - Algoritma yang didasari Teorema Bayes, kemungkinan setiap fitur muncul dalam suatu kelas, untuk klasifikasi teks
7. Neural Networks
