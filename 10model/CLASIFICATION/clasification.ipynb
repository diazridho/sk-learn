## <font color='#c8cd34' face='Courier New'>ALGORITMA KLASIFIKASI</font>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    1. KNN (K Nearest Neighbors) <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Mengklasifikasikan data berdasarkan kedekatan dengan data yang sudah diberi label. Cocok untuk data yang memiliki pola berkelompok <br>
    <br>
    2. Decision Tree <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Dapat mengelola data tidak terstruktur, mudah divisualisasikan, tidak mengasumsikan distribusi/skala. <br>
    <br>
    3. Random Forest <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Versi lanjutan (Ensemble) yang menggabungkan banyak Decision Tree. <br>
    <br>
    4. Logistic Regression <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Mengubah persamaan linear menjadi probabilitas untuk klasifikasi biner. Untuk klasifikasi binner (Ya/Tidak) <br>
    <br>
    5. Support Vector Machine (SVM) <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Menemukan garis pemisah (hyperplane) untuk membagi data ke kelas tertentu. Efektif untuk data yang terpisah jelas. Untuk klasifikasi binner/Multikelas <br>
    <br>
    6. Naive Bayes <br>
    &nbsp;&nbsp;&nbsp;&nbsp;- Berdasarkan Teorema Bayes, sangat efisien untuk klasifikasi teks dan banyak fitur kategori. <br>
</div>
