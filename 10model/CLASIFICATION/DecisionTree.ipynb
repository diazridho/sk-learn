## <font color='#00eeff'>Parameter Decision Tree</font>

### <font color='#00eeff'>Kriteria (default='gini')</font>

- Menilai kualitas dari pembagian data dalam pohon keputusan, menentukan fitur mana yang paling dibutuhkan dengan membandingkan nilai gini/entropi 

1. GINI IMPURITY ( 1 - sum P^2). lebih cepat, untuk dataset besar
2. ENTROPI IMPURITY (- sum p LOG2 (p)). lebih teliti, untuk analisis

### <font color='#00eeff'>max_depth (default=None)</font>

Jika dibatasi dapat mencegah overfitting

### <font color='#00eeff'>min_sample_split (default=2)</font>

Minimal jumlah data yang memperbolehkan node untuk melakukan split.
2 adalah paling agresif

### <font color='#00eeff'>min_sample_leaf (default=1)</font>

Minimal jumlah sampel yang harus ada di leaf node.

### <font color='#00eeff'>max_features (default= None)</font>

Menentukan berapa banyak fitur yang dipertimbangkan untuk dipakai dalam tiap node. Output tiap node tetap 1 fitur yang akan dipilih

- None (Dilakukan pengecekan seluruh fitur)
- int/float (Dipilih secara acak fitur yang dipertimbangkan)
- sqrt/log (100 fitur akan dipilih 10 fitur jika sqrt)

## <font color='#ccff00'>Ciri-ciri Decision Tree</font>

1. Tidak mengasumsikan distribusi dan skala
2. Penafsiran mudah untuk visualisasi
3. Sensitif terhadap noise, terutama jika parameter-parameternya di-set terlalu spesifik
4. Bisa untuk segala jenis klasifikasi
