## <font color='#34cacd' face='Courier New'>Parameter Decision Tree</font>

### <font color='#34cacd' face='Courier New'>Kriteria (default='gini')</font>

<div style="color:#ffffff; font-family:Courier New;">
    Menilai kualitas dari pembagian data dalam pohon keputusan, menentukan fitur mana yang paling dibutuhkan dengan membandingkan nilai gini/entropi  <br> <br>
</div>
<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    - GINI IMPURITY ( 1 - sum P^2). lebih cepat, untuk dataset besar <br>
    - ENTROPI IMPURITY (- sum p LOG2 (p)). lebih teliti, untuk analisis
</div>

### <font color='#34cacd' face='Courier New'>max_depth (default=None)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Jika dibatasi dapat mencegah overfitting <br>
</div>

### <font color='#34cacd' face='Courier New'>min_sample_split (default=2)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Minimal jumlah data yang memperbolehkan node untuk melakukan split. 2 adalah paling agresif <br>
</div>

### <font color='#34cacd' face='Courier New'>min_sample_leaf (default=1)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Minimal jumlah sampel yang harus ada di leaf node. <br>
</div>

### <font color='#34cacd' face='Courier New'>max_features (default= None)</font>

<div style="color:#ffffff; font-family:Courier New;">
    Menentukan berapa banyak fitur yang dipertimbangkan untuk dipakai dalam tiap node. Output tiap node tetap 1 fitur yang akan dipilih <br> <br>
</div>
<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    - None (Dilakukan pengecekan seluruh fitur) <br>
    - int/float (Dipilih secara acak fitur yang dipertimbangkan) <br>
    - sqrt/log (100 fitur akan dipilih 10 fitur jika sqrt)
</div>

## <font color='#34cacd' face='Courier New'>Ciri-ciri Decision Tree</font>

<div style="background-color:#1E3D59; color:#ffffff; font-family: 'Courier New', monospace; padding: 20px; line-height: 1.5;">
    1. Tidak mengasumsikan distribusi dan skala <br>
    2. Penafsiran mudah untuk visualisasi <br>
    3. Sensitif terhadap noise, terutama jika parameter-parameternya di-set terlalu spesifik <br>
    4. Bisa untuk segala jenis klasifikasi

</div>
