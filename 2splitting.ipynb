```python
from sklearn.datasets import load_iris, make_blobs
from sklearn.model_selection import train_test_split

data = load_iris()
X = data.data
y = data.target
```

```python
import numpy as np
import matplotlib.pyplot as plt

#train test split biasa
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2) #test size 20%

counts = np.bincount(y_train)
positions = np.arange(3)
plt.bar(positions, counts)
plt.xticks(positions, data.target_names)
```

```python
from sklearn.model_selection import StratifiedShuffleSplit

# stratified splitting (memastikan pembagian sampel tiap kelas seimbang)
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(X,y):
    X_train, X_test= X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
```
