```python
from sklearn.datasets import fetch_openml
import pandas as pd

data = fetch_openml('car', as_frame=True, version=2).frame
data
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories= [['big', 'med', 'small'],  # untuk kolom pertama (buying)
                ['high', 'low', 'med']]  # for 'buying' and 'maint' columns
)
col_to_encode = ['lug_boot', 'safety']
data[col_to_encode] = encoder.fit_transform(data[col_to_encode])
data
```

```python
# Tujuan untuk membuat menambah kolom baru yang berisi encode dari kategori buying dan maint
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

col_to_enc = ['buying','maint']
encoder.fit(data[col_to_enc]) # Memberikan data train ke encoder

encoded_array = encoder.transform(data[col_to_enc]) #memproses data yang sudah diberikan
feature_names = encoder.get_feature_names_out(col_to_enc)

df_encoded = pd.DataFrame(encoded_array, columns=feature_names, index=data.index) #datframe (isi, namakolom)

df_final = pd.concat([ #menyatukan, axis 1 untuk menambahkan dari kanan
    data.drop(columns=col_to_enc),
    df_encoded
], axis=1)
df_final
```
